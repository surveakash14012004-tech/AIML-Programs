```python
import math

print("Square root:", math.sqrt(25))
print("Power:", math.pow(2, 3))
print("Factorial:", math.factorial(5))
print("Value of pi:", math.pi)
```

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Array:", numbers)
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
print("Standard Deviation:", np.std(numbers))
```

```python
from scipy import stats

numbers = [10, 20, 30, 40, 50]

print("Mean:", stats.tmean(numbers))
print("Standard Deviation:", stats.tstd(numbers))
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Akash"],
    "Marks": [85, 78, 92, 88, 95]
}

df = pd.DataFrame(data)

print(df)
print("\nAverage Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
```

```python
import matplotlib.pyplot as plt

names = ["Amit", "Rahul", "Priya", "Sneha", "Akash"]
marks = [85, 78, 92, 88, 95]

plt.bar(names, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

names = ["Amit", "Rahul", "Priya", "Sneha", "Akash"]
marks = [85, 78, 92, 88, 95]

sns.barplot(x=names, y=marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks using Seaborn")
plt.show()
```
