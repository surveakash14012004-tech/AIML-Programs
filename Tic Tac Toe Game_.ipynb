```python


board = [" " for _ in range(9)]


def display():
  print(board[0] + "|" + board[1] + "|" + board[2])
  print("-+-+-")
  print(board[3] + "|" + board[4] + "|" + board[5])
  print("-+-+-")
  print(board[6] + "|" + board[7] + "|" + board[8])
  print()

def check_winner(player):
  win = [
      (0,1,2), (3,4,5), (6,7,8),
      (0,3,6), (1,4,7), (2,5,8),
      (0,4,8), (2,4,6)
  ]

  for a, b, c in win:
    if board[a] == board[b] == board[c] == player:
      return True
  return False

def ai_move():
  #try to win
  for i in range(9):
    if board[i] == " ":
      board[i] = "O"
      if check_winner("O"):
        return
      board[i] = " "

  #block the player
  for i in range(9):
    if board[i] == " ":
      board[i] = "X"
      if check_winner("X"):
        board[i] = "O"
        return
      board[i] = " "

  #choose first available postion
  for i in range(9):
    if board[i] == " ":
      board[i] = "O"
      return

#Main program
print("TIC-TAC-TOE")
print("You are X, Computer is O")

while True:
  display()

  #User move
  position = int(input("Enter position (1-9): ")) - 1

  if position < 0 or position > 8 or board[position] !=" ":
    print("Invalid move!")
    continue

  board[position] = "X"

  if check_winner("X"):
    display()
    print("You Win!")
    break

  if " " not in board:
    display()
    print("Draw!")
    break

  #AI move
  ai_move()

  if check_winner("O"):
    display()
    print("Computer Wins!")
    break

  if " " not in board:
    display()
    print("Draw!")
    break
```
