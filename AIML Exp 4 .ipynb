```python
import statistics
```

```python
data = [20, 40, 60, 80, 100, 120, 140]
```

```python
mean = statistics.mean(data)
```

```python
median = statistics.median(data)
```

```python
std_dev = statistics.stdev(data)
```

```python
print("Data:", data)
print("Mean =", mean)
print("Median =", median)
print("Standard Deviation =", std_dev)
```
